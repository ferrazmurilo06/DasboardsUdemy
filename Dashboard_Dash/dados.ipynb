```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

# Carregando o Dataset do Titanic.

```python
url = "https://web.stanford.edu/class/archive/cs/cs109/cs109.1166/stuff/titanic.csv"
dados = pd.read_csv(url)
dados
```

# Histograma de idade dos passageiros

```python
plt.hist(dados["Age"].dropna(), bins=30)
plt.title("Faixa Etária")
plt.xlabel("Idade")
plt.ylabel("Frequência")
```

```python
dados['Survived'].value_counts()# 0 - não sobreviveu / 1 - sobreviveu
```

# Boxplot da idade dos sobreviventes

```python
sns.boxplot(y="Age", x="Survived", data=dados, hue="Survived")
plt.title("Idade por Sobrevivência")
plt.xlabel("Sobrevivência (0 = Não | 1 = Sim)")
plt.ylabel("Idade")
```

# Histograma da idade

```python
fig = px.histogram(dados, x="Age", title="Distribuição de Idade")
fig.show()
```
